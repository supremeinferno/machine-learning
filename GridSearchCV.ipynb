```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset('iris')
df
```

```python
df['species'].unique()
```

```python
from sklearn.model_selection import train_test_split
X = df.drop('species',axis=1)
y = df['species']
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.33,random_state=42)
```

# KNN

```python
from sklearn.neighbors import KNeighborsClassifier
model_knn = KNeighborsClassifier(n_neighbors=5)
model_knn.fit(X_train,y_train)
```

```python
model_knn.score(X_test,y_test)
```

# SVC

```python
from sklearn.svm import SVC
#model_svc = SVC(kernel='rbf', gamma='auto', C=30)
model_svc = SVC(kernel='rbf')
model_svc.fit(X_train,y_train)
```

```python
model_svc.score(X_test,y_test)
```

# lets use the Grid Search CV

model_svc

```python
from sklearn.model_selection import GridSearchCV
Classifier = GridSearchCV(model_svc, {
    'gamma': ['auto'],
    'C': [1, 10, 20, 25, 30],
    'kernel': ['linear', 'rbf']
    }, 
    cv=5, return_train_score=False)
```

```python
Classifier.fit(X, y)
```

```python
result = pd.DataFrame(Classifier.cv_results_)
result
```

```python
result[['param_C', 'param_kernel', 'mean_test_score']]
```

```python
result.loc[[result['mean_test_score'].idxmax()]]
```

```python
result.nlargest(1, 'mean_test_score')
```

```python
result[result['mean_test_score'] == result['mean_test_score'].max()]
```

model_knn

```python
from sklearn.model_selection import GridSearchCV
Classifier2 = GridSearchCV(model_knn, {
    'algorithm': ['auto'],
    'n_neighbors': [5, 10, 15, 25],
    'metric': ['euclidean', 'manhattan', 'minkowski']
    }, 
    cv=5, return_train_score=False)
```

```python
Classifier2.fit(X, y)
Classifier2
```

```python
result2 = pd.DataFrame(Classifier2.cv_results_)
result2
```

```python
result2[['param_n_neighbors', 'param_metric', 'mean_test_score']].sort_values(by='mean_test_score', ascending=False)
```

```python
result2[result2['mean_test_score'] == result2['mean_test_score'].max()]
```
